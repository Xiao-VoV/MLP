这个四维张量可以看作是由多个三维张量组成，而每个三维张量又由多个二维张量组成，每个二维张量由多个一维向量组成。
高维张量在多个领域都有着极其重要的实际意义，
图像识别
在图像数据处理中，彩色图像通常被表示为三维张量，形状为 [高度, 宽度, 通道数]，其中通道数一般为 3（分别对应红、绿、蓝三个颜色通道）。而在处理批量图像时，就会使用到四维张量，形状为 [批量大小, 高度, 宽度, 通道数]。例如在训练卷积神经网络（CNN）进行图像分类任务时，会将多张图像组成一个批量输入到网络中，这里的批量数据就是四维张量。
当处理视频数据时，由于视频是由一系列连续的图像帧组成，此时就需要用到五维张量，形状为 [批量大小, 帧数, 高度, 宽度, 通道数]。

## tensor 操作

````python
# Shapes need to be in the right way  
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11], 
                         [9, 12]], dtype=torch.float32)
# mm是矩阵乘法的缩写
torch.mm(tensor_A, tensor_B.T)
````

````python
# Since the linear layer starts with a random weights matrix, let's make it reproducible (more on this later)
torch.manual_seed(42)
# This uses matrix multiplication
linear = torch.nn.Linear(in_features=2, # in_features = matches inner dimension of input 
                         out_features=10) # out_features = describes outer value 
x = tensor_A
output = linear(x)
print(f"Input is_grad:{x.requires_grad} shape: {x.shape}\n")
print(f"Output:\n{output}\n\nOutput shape: {output.shape}")
print(f"wights shape: {linear.weight}")
````
